# Validações Estatísticas do BeamNG como Digital Twin

## 1. Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from os import listdir
import ast

from scipy.signal import correlate, savgol_filter
from scipy.stats import pearsonr
```

## 2. Carregamento dos Dados

```python
simulados = []

for file in listdir('circuito4/simulado_taxa_coleta_correta'):
    df = pd.read_csv(f'circuito4/simulado_taxa_coleta_correta/{file}', sep=';')
    
    # Converter velocidade (vetor → km/h)
    df['vel'] = df['vel'].apply(
        lambda x: np.sqrt(sum(i**2 for i in ast.literal_eval(x))) * 3.6
    )
    
    # Distância acumulada
    dx = df["posX"].diff()
    dy = df["posY"].diff()
    dz = df["posZ"].diff()

    df["dist_acumulada_km"] = np.sqrt(dx**2 + dy**2 + dz**2).cumsum() / 1000
    
    simulados.append(df)
```

## 3. Padronização de Colunas

```python
map_cols = {
    'time': 'Time (sec)',
    'throttle': ' Posição D do pedal do acelerador (%)',
    'vel': ' Velocidade do veículo (km/h)',
    'posY': ' Latitude (deg)',
    'posX': ' Longitude (deg)',
    'posZ': ' Altitude (m)',
    'dist_acumulada_km': ' Distância da viagem (km)'
}

for sim in simulados:
    sim.rename(columns=map_cols, inplace=True)

colunas = list(map_cols.values())
```

## 4. Alinhamento Temporal

```python
def preparar_df(df, time_col='Time (sec)'):
    df = df.copy()
    
    df.columns = df.columns.str.strip()
    df = df.sort_values(time_col)
    
    # Converter tempo (segundos → timedelta)
    df[time_col] = pd.to_timedelta(df[time_col], unit='s')
    
    df = df.set_index(time_col)
    
    # Garantir numérico
    df = df.apply(pd.to_numeric, errors='coerce')
    
    # Reamostragem (10 Hz)
    df = df.resample('100ms').mean().interpolate()
    
    return df
```

### Real

```python
def pipeline():
    
    # -------- SIMULADO --------
    simulados = []
    for file in listdir('circuito4/simulado_taxa_coleta_correta'):
        df = pd.read_csv(f'circuito4/simulado_taxa_coleta_correta/{file}', sep=';')
        
        # velocidade (vetor → km/h)
        df['vel'] = df['vel'].apply(
            lambda x: np.sqrt(sum(i**2 for i in ast.literal_eval(x))) * 3.6
        )
        
        simulados.append(df)
# -------- REAL --------
    reais = []
    for file in listdir('circuito4/real'):
        df = pd.read_csv(f'circuito4/real/{file}', sep=',')
        reais.append(df)
    
    # -------- PREPARAR --------
    real = preparar_df(reais[0])
    simulado = preparar_df(simulados[0])
    
    # -------- LIMPAR NOMES --------
    real.columns = real.columns.str.strip()
    simulado.columns = simulado.columns.str.strip()
    
    # -------- JUNTAR --------
    df = pd.concat([
        real.add_prefix('real_'),
        simulado.add_prefix('sim_')
    ], axis=1).dropna()
    
    # -------- VARIÁVEL --------
    col = 'Velocidade do veículo (km/h)'
```

## 5.  Sincronização + Lag

```python
x = df[f'real_{col}'].values
    y = df[f'sim_{col}'].values
    
    corr = correlate(x, y)
    lag = np.argmax(corr) - len(x)
    
    print("Lag detectado:", lag)
    
    df['sim_shifted'] = df[f'sim_{col}'].shift(lag)
    df = df.dropna()
```

## 6. Suavização

```python
df['real_smooth'] = savgol_filter(df[f'real_{col}'], 11, 2)
df['sim_smooth']  = savgol_filter(df['sim_shifted'], 11, 2)
```

## 7. Normalização LOCAL 

```python
window = 50
    
    def zscore(x):
        return (x - x.mean()) / x.std()
    
    df['real_norm'] = df['real_smooth'].rolling(window).apply(lambda x: zscore(x)[-1])
    df['sim_norm']  = df['sim_smooth'].rolling(window).apply(lambda x: zscore(x)[-1])
    
    # -------- CORRELAÇÃO --------
    df['corr'] = df['real_norm'].rolling(window).corr(df['sim_norm'])
    
    return df
```

## 8. Correlação com Janela Deslizante

```python
df = pipeline()
```

## 9. Gráficos
- Séries alinhadas

```python
# Séries normalizadas
plt.figure(figsize=(12,5))
plt.plot(df.index, df['real_norm'], label='Real')
plt.plot(df.index, df['sim_norm'], label='Simulado')
plt.legend()
plt.title('Séries Normalizadas (Alinhadas)')
plt.show()
```

* Correlação ao longo do tempo

```python
# Correlação
plt.figure(figsize=(12,4))
plt.plot(df.index, df['corr'], label='Correlação')
plt.axhline(0, linestyle='--')
plt.legend()
plt.title('Correlação de Pearson (Janela Deslizante)')
plt.show()
```

## 10. Pearson Global 

```python

# Pearson global
r, p = pearsonr(df['real_norm'].dropna(), df['sim_norm'].dropna())
print(f"Pearson global: {r:.3f}")
print(f"p-value: {p:.6f}")
```
